```python
import numpy as np
import torch
import torch.nn as nn
```

```python
class Model(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(5,5,1)

    def forward(self, x):
        x = self.conv1(x)
        return x


yoloprev=Model()
diffusion=Model()
yoloaft=Model()
for param in yoloaft.parameters():
    param.requires_grad=False
```

```python

x_leaf = torch.ones(1,5,1,1, requires_grad = True)
x_blur_leaf = torch.ones(1,5,1,1, requires_grad = True)+1

with torch.no_grad():
    x_latent = yoloprev(x_leaf)
    x_blur_latent = yoloprev(x_blur_leaf)
    
x_blur_rev = diffusion(x_blur_latent)

loss_latent = (x_latent - x_blur_rev).abs().mean()
print(loss_latent)
    

loss_latent.backward()
print(x_leaf.requires_grad)
print(x_blur_leaf.requires_grad)
print('loss_latent.grad_fn = '+str(loss_latent.grad_fn))
print('x_leaf.grad = '+str(x_leaf.grad))
print('x_blur_leaf.grad = '+str(x_blur_leaf.grad))



print('-----yoloprev-----')
for name, param in yoloprev.named_parameters():
    print(name)
    #print(param)
    print(param.requires_grad)
    print(param.grad)
    
print('-----diffusion-----')
for name, param in diffusion.named_parameters():
    print(name)
    #print(param)
    print(param.requires_grad)
    print(param.grad)
    
print('-----yoloaft-----')
for name, param in yoloaft.named_parameters():
    print(name)
    #print(param)
    print(param.requires_grad)
    print(param.grad)
#backward()
#print('data.grad = '+str(data.grad))
#print('x1.grad = '+str(x1.grad))
#print('x2.grad = '+str(x2.grad))
#print('x3.grad = '+str(x3.grad))
#print('z.grad = '+str(z.grad))
```
